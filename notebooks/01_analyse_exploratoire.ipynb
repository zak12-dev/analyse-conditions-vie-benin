```python
cles_menage = ["grappe", "menage", "vague"]

# Vérifier que chaque ménage existe dans le fichier ménage
cles_reference = df_menage[cles_menage].drop_duplicates()

for nom, df in {
    "individu": df_individu,
    "consommation": df_conso,
    "ponderations": df_ponderations
}.items():

    cles_fichier = df[cles_menage].drop_duplicates()

    comparaison = cles_fichier.merge(
        cles_reference,
        on=cles_menage,
        how="left",
        indicator=True
    )

    print(f"\n--- {nom.upper()} ---")
    print("Ménages présents :", len(cles_fichier))
    print(
        "Ménages correspondants :",
        (comparaison["_merge"] == "both").sum()
    )
    print(
        "Ménages non retrouvés :",
        (comparaison["_merge"] == "left_only").sum()
    )
```

```python
cles = ["grappe", "menage", "vague"]

# Référence : fichier des ménages
reference = df_menage[cles + ["hhid"]].drop_duplicates()

for nom, df in {
    "individu": df_individu,
    "consommation": df_conso,
    "ponderations": df_ponderations
}.items():

    comparaison = reference.merge(
        df[cles + ["hhid"]].drop_duplicates(),
        on=cles,
        how="inner",
        suffixes=("_reference", "_fichier")
    )

    identiques = (
        comparaison["hhid_reference"]
        == comparaison["hhid_fichier"]
    ).sum()

    differents = (
        comparaison["hhid_reference"]
        != comparaison["hhid_fichier"]
    ).sum()

    print(f"\n--- {nom.upper()} ---")
    print("Identifiants identiques :", identiques)
    print("Identifiants différents :", differents)
```

```python
resultats_contrastes = []

for variable in variables_conditions_vie:

    estimation = echantillon.estimation.mean(
        y=variable,
        by="milieu_residence"
    )

    contraste = estimation.contrast(
        svy.estd("Urbain") - svy.estd("Rural")
    )

    resultat = contraste.to_polars().to_dicts()[0]

    resultats_contrastes.append({
        "variable": variable,
        "difference_pp": resultat["est"] * 100,
        "erreur_type_pp": resultat["se"] * 100,
        "borne_inf_95": resultat["lci"] * 100,
        "borne_sup_95": resultat["uci"] * 100,
        "statistique_t": resultat["t"],
        "p_value": resultat["p_value"],
        "degres_liberte": resultat["df"]
    })

df_contrastes = pd.DataFrame(resultats_contrastes)

df_contrastes["significatif_5pct"] = (
    df_contrastes["p_value"] < 0.05
)

df_contrastes = df_contrastes.sort_values(
    "difference_pp",
    ascending=False
).reset_index(drop=True)

display(df_contrastes.round(4))
```

```python
resultats_contrastes = []

for variable in variables_conditions_vie:
    estimation = echantillon.estimation.mean(
        y=variable,
        by="milieu_residence"
    )

    contraste = estimation.contrast(
        svy.estd("Urbain") - svy.estd("Rural")
    )

    resultat = contraste.to_polars().to_dicts()[0]

    resultats_contrastes.append({
        "variable": variable,
        "difference_pp": resultat["est"] * 100,
        "erreur_type_pp": resultat["se"] * 100,
        "borne_inf_95": resultat["lci"] * 100,
        "borne_sup_95": resultat["uci"] * 100,
        "statistique_t": resultat["t"],
        "p_value": resultat["p_value"],
        "degres_liberte": resultat["df"]
    })

df_contrastes = pd.DataFrame(resultats_contrastes)

df_contrastes["significatif_5pct"] = (
    df_contrastes["p_value"] < 0.05
)

df_contrastes = df_contrastes.sort_values(
    "difference_pp",
    ascending=False
).reset_index(drop=True)

display(df_contrastes.round(4))
```

```python
controle_coherence = []

for variable in variables_conditions_vie:
    taux_national = df_nationaux.loc[
        df_nationaux["variable"] == variable,
        "taux_national_pct"
    ].iloc[0]

    donnees = df_resultats_departements[
        df_resultats_departements["variable"] == variable
    ].copy()

    poids_departements = (
        df_analyse
        .groupby("s00q01")["poids"]
        .sum()
    )

    donnees["poids_total"] = donnees["departement"].map(
        poids_departements
    )

    taux_reconstitue = np.average(
        donnees["taux_pct"],
        weights=donnees["poids_total"]
    )

    controle_coherence.append({
        "variable": variable,
        "taux_national_pct": taux_national,
        "taux_reconstitue_pct": taux_reconstitue,
        "ecart_pp": taux_reconstitue - taux_national
    })

df_controle_coherence = pd.DataFrame(controle_coherence)

display(df_controle_coherence.round(6))
```
